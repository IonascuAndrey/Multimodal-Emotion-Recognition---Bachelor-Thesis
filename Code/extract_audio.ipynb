```python
import os
import subprocess
from pathlib import Path
from tqdm import tqdm

# -------- CONFIG --------
INPUT_ROOT = Path("../Data/MELD.Raw")
OUTPUT_ROOT = Path("../Data/MELD.Audio")  # new folder

SPLITS = ["train", "dev", "test"]

# create output folders
for split in SPLITS:
    (OUTPUT_ROOT / split).mkdir(parents=True, exist_ok=True)


def extract_audio(video_path, audio_path):
    """
    Extract audio using ffmpeg:
    - mono (-ac 1)
    - 16kHz (-ar 16000)
    - overwrite (-y)
    """
    cmd = [
        "ffmpeg",
        "-y",
        "-i", str(video_path),
        "-ac", "1",
        "-ar", "16000",
        str(audio_path)
    ]
    
    # suppress output for speed
    subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)


# -------- PROCESS --------
for split in SPLITS:
    if split == "train":
        video_dir = INPUT_ROOT / "train" / "train_splits"
    elif split == "dev":
        video_dir = INPUT_ROOT / "dev" / "dev_splits_complete"
    else:
        video_dir = INPUT_ROOT / "test" / "output_repeated_splits_test"

    audio_dir = OUTPUT_ROOT / split

    videos = list(video_dir.glob("*.mp4"))

    print(f"{split}: {len(videos)} files")

    for v in tqdm(videos):
        a = audio_dir / (v.stem + ".wav")

        # skip if already exists (important, don't redo work)
        if a.exists():
            continue

        extract_audio(v, a)
```
